# Attention Topology

Calculate which source positions can reach each token across attention layers.

This is the full **PyTorch** version of a challenge from [LLM Quest](https://bankoti.github.io/llm-quest). The in-browser challenge grades numpy; here you work with real tensors. Fill in each `TODO`, then run the checks cell.

Runs on the free Colab CPU runtime; switch to a GPU via *Runtime > Change runtime type* if you want to experiment at scale. PyTorch comes preinstalled.

```python
from typing import Optional


def one_layer_reach(length: int, window: Optional[int]) -> list[set[int]]:
    result = []
    for query in range(length):
        start = 0 if window is None else max(0, query - window + 1)
        result.append(set(range(start, query + 1)))
    return result


def compose(previous: list[set[int]], current: list[set[int]]) -> list[set[int]]:
    return [{source for parent in parents for source in previous[parent]} for parents in current]


length = 8
local = one_layer_reach(length, window=3)
two_local_layers = compose(local, local)
```

```python
assert two_local_layers[-1] == {3, 4, 5, 6, 7}

global_layer = one_layer_reach(length, window=None)
local_then_global = compose(local, global_layer)
assert local_then_global[-1] == set(range(length))
print("Architecture lab 02 passed")
```
