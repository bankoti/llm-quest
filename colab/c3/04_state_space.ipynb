```python
import torch

from mini_llm.architectures import SelectiveStateSpaceLayer


torch.manual_seed(11)
width, state_size = 8, 4
layer = SelectiveStateSpaceLayer(width, state_size).eval()
inputs = torch.randn(2, 6, width)
changed_future = inputs.clone()
changed_future[:, 4:] += 100.0
with torch.no_grad():
    output = layer(inputs)
    changed = layer(changed_future)
```

```python
assert torch.allclose(output[:, :4], changed[:, :4], atol=1e-6)

# Generation state per layer is proportional to batch * width * state_size,
# independent of the number of tokens already processed.
state_elements = inputs.size(0) * width * state_size
assert state_elements == 64
print("Architecture lab 04 passed")
```
