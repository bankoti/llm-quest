## A small finite world with a hidden rule

In addition modulo p, the answer to a+b is the remainder after division by p. For p=5, 4+3=7 becomes 2. There are p squared ordered input pairs. Split pairs before training and never move a pair between train and test after viewing the curve. A network can memorize observed pairs without learning an algorithm for unseen pairs.

Grokking names a delayed transition to generalization after fitting training data. It is an empirical phenomenon, not a mandatory phase of every neural network. Regularization, dataset fraction, architecture, learning rate, seed, and training duration all affect whether and when it appears. A short CPU run is a smoke test and must not be labeled a reproduction of a long delayed transition.

## Clock features express a reusable rule

Map a to angle theta_a=2*pi*a/p. The addition identities give cos(theta_a+theta_b)=cos(theta_a)cos(theta_b)-sin(theta_a)sin(theta_b), and sin(theta_a+theta_b)=sin(theta_a)cos(theta_b)+cos(theta_a)sin(theta_b). Match the resulting point to the p candidate clock points by dot product. The maximum is the modular sum, up to numerical ties.

The Nanda et al. case study analyzed a one-layer Transformer on modular addition and identified Fourier-based computations. Our clock decoder is an explicit constructive solution, while the training experiment uses a small neural network and measures its actual held-out behavior. Finding a successful hand-built circuit does not prove a trained network uses it; that requires activation analysis and interventions in the next unit.

## Worked Example: Wrap around a five-hour clock

a=4, b=3, p=5.

1. **Angles:** The angles add to 14*pi/5, equivalent modulo 2*pi to 4*pi/5.
2. **Decode:** Candidate 2 has angle 4*pi/5, so its dot product with the sum feature is 1.
3. **Compare learning:** A model trained on only some pairs must discover a transferable rule or another generalizing representation to answer withheld pairs.

## Worked Solution

Read the explanation and docstrings alongside the code. Verify nonmutation, masks, normalization, and numerical conventions against your own implementation.


````python
"""Constructive clock circuit, separate from learning the circuit."""
import numpy as np


def clock_sum(a, b, modulus):
    """Return (2D sum feature, decoded ID) via trigonometric addition.

    a,b are integers in [0,modulus), modulus integer >=2. Invalid: ValueError.
    Feature is [cos(theta_a+theta_b), sin(theta_a+theta_b)], constructed from
    separate sin/cos of a and b. Decode by candidate clock dot products.
    """
    if any(not isinstance(v, int) or isinstance(v, bool) for v in (a, b, modulus)) or modulus < 2 or not 0 <= a < modulus or not 0 <= b < modulus:
        raise ValueError('invalid modular operands')
    aa, bb = 2*np.pi*a/modulus, 2*np.pi*b/modulus
    feature = np.array([np.cos(aa)*np.cos(bb)-np.sin(aa)*np.sin(bb), np.sin(aa)*np.cos(bb)+np.cos(aa)*np.sin(bb)])
    angles = 2*np.pi*np.arange(modulus)/modulus
    candidates = np.stack([np.cos(angles), np.sin(angles)], axis=1)
    return feature, int(np.argmax(candidates @ feature))
````

````python
import numpy as np
for p in [3, 5, 7, 11]:
    for a in range(p):
        for b in range(p):
            feature, result = clock_sum(a, b, p)
            angle = 2*np.pi*((a+b) % p)/p
            assert np.allclose(feature, [np.cos(angle), np.sin(angle)])
            assert result == (a+b) % p
try:
    clock_sum(1, 2, 1)
except ValueError:
    pass
else:
    raise AssertionError('Invalid modulus must fail')
print('Clock-circuit checks passed.')
````

## Measured Experiment

Predict before running. Change one variable at a time. Report actual outcomes and failures; the lesson does not promise a successful learning curve for every configuration.


````python
"""Actual modular training, frozen probes, and frequency interventions.

Default 300 updates is a smoke run, NOT a promised grokking reproduction.
To investigate longer: MODULAR_STEPS=20000 python experiments/dynamics_modular.py
Reports all seeds; no external data. This MLP differs from the paper Transformer.
"""
import json
import os
import numpy as np
import torch
from torch import nn
from torch.nn import functional as F


def modular_experiment(steps=300):
    torch.set_num_threads(1)
    modulus = 17
    pairs = torch.cartesian_prod(torch.arange(modulus), torch.arange(modulus))
    inputs = torch.cat([F.one_hot(pairs[:, 0], modulus), F.one_hot(pairs[:, 1], modulus)], -1).float()
    targets = pairs.sum(-1) % modulus
    split = torch.randperm(len(pairs), generator=torch.Generator().manual_seed(44))
    train, test = split[:100], split[100:]
    reports = []
    for seed in [3, 8, 21]:
        torch.manual_seed(seed)
        hidden = nn.Sequential(nn.Linear(2*modulus, 64), nn.ReLU())
        head = nn.Linear(64, modulus)
        model = nn.Sequential(hidden, head)
        optimizer = torch.optim.AdamW(model.parameters(), lr=.003, weight_decay=.1)
        curve = []
        for step in range(steps):
            optimizer.zero_grad(set_to_none=True)
            loss = F.cross_entropy(model(inputs[train]), targets[train]); loss.backward(); optimizer.step()
            if step % max(1, steps//10) == 0 or step == steps-1:
                with torch.no_grad():
                    logits = model(inputs)
                    curve.append(dict(step=step+1, train_accuracy=float((logits[train].argmax(-1)==targets[train]).float().mean()), test_accuracy=float((logits[test].argmax(-1)==targets[test]).float().mean())))
        with torch.no_grad():
            features = hidden(inputs).numpy()
            logits = model(inputs).numpy().astype(np.float64).reshape(modulus, modulus, modulus)
        onehot = np.eye(modulus)[targets.numpy()]
        h = features[train]; y = onehot[train]
        weights = np.linalg.solve(h.T@h + np.eye(64), h.T@y)
        probe_accuracy = float((features[test]@weights).argmax(-1).__eq__(targets[test].numpy()).mean())
        shuffled = np.random.default_rng(seed).permutation(y)
        shuffled_weights = np.linalg.solve(h.T@h + np.eye(64), h.T@shuffled)
        shuffled_accuracy = float(((features[test]@shuffled_weights).argmax(-1)==targets[test].numpy()).mean())
        spectrum = np.fft.fft2(logits, axes=(0, 1))
        energy = np.abs(spectrum).sum(-1)**2
        # Predeclared illustrative diagonal frequencies, not a discovered circuit.
        mask = np.eye(modulus, dtype=bool)
        restricted = np.fft.ifft2(np.where(mask[..., None], spectrum, 0), axes=(0, 1)).real
        excluded = logits - restricted
        def heldout_loss(values):
            return float(F.cross_entropy(torch.tensor(values.reshape(-1, modulus)[test]), targets[test]))
        assert np.allclose(restricted+excluded, logits)
        reports.append(dict(seed=seed, steps=steps, train_pairs=len(train), test_pairs=len(test), curve=curve,
            probe_accuracy=probe_accuracy, shuffled_probe_accuracy=shuffled_accuracy,
            total_loss=heldout_loss(logits), restricted_loss=heldout_loss(restricted), excluded_loss=heldout_loss(excluded),
            diagonal_frequency_energy_fraction=float(energy[mask].sum()/energy.sum()),
            neuron0_grid=features[:, 0].reshape(modulus, modulus).round(3).tolist(),
            delayed_transition_observed=any(p['train_accuracy']>.95 and p['test_accuracy']<.3 for p in curve[:-1]) and curve[-1]['test_accuracy']>.9))
    return reports


if __name__ == '__main__':
    print(json.dumps(modular_experiment(int(os.environ.get('MODULAR_STEPS', '300'))), indent=2))
````
